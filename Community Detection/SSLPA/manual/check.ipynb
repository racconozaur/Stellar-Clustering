```python
import pandas as pd
```

```python
def print_res(path, TYPE, check):
    df = pd.read_csv(path)

    label_counts = df['label'].value_counts()

    print(f"\nTOP 10 LARGEST LABELS in {TYPE}")
    print(label_counts.head(10))

    print("------------------------")
    print("check for: ", check)
    print(label_counts[check])
    print("------------------------")

    print("\n=== BOTTOM 10 SMALLEST LABELS ===")
    print(label_counts.tail(10))


    print(f"\nSUMMARY")
    print(f"Labels with only 1 node: {sum(label_counts == 1)}")
    print(f"Labels with > 1000 nodes: {sum(label_counts > 1000)}")
    print(f"Labels with > 100,000 nodes: {sum(label_counts > 100000)}")


    print("\n=== DOMINANT LABEL ===")
    top_label = label_counts.index[0]
    top_count = label_counts.iloc[0]
    print(f"{top_label}: {top_count:,} nodes ({top_count/len(df)*100:.1f}% of graph)")
```

```python
print_res("normalized/sslpa_tx_lcc_manual_labels.csv",
          "normalized",
          "UNLABELED"
          )
```

# Raw

```python
print_res("raw/sslpa_tx_lcc_manual_labels.csv",
          "raw",
          "UNLABELED"
          )
```
